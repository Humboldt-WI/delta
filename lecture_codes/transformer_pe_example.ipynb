```python
import numpy as np
import matplotlib.pyplot as plt

def positional_encoding(max_len, d_model):
    # Create matrix of [max_len, d_model] shape
    PE = np.zeros((max_len, d_model))
    
    # Compute position indices (max_len x 1)
    position = np.arange(max_len)[:, np.newaxis]
    
    # Compute dimension indices (1 x d_model) → for 2i
    div_term = np.exp(np.arange(0, d_model, 2) * (-np.log(10000.0) / d_model))
    
    # Apply sin to even indices (2i)
    PE[:, 0::2] = np.sin(position * div_term)
    
    # Apply cos to odd indices (2i + 1)
    PE[:, 1::2] = np.cos(position * div_term)
    
    return PE

# Example: 50 positions, model dimension 16
max_len = 100
d_model = 20
PE = positional_encoding(max_len, d_model)

# Plot positional encodings for first few dimensions
plt.figure(figsize=(10, 6))
for i in np.arange(4, 8, 1):
    plt.plot(PE[:, i], label=f'dim {i}')
plt.legend()
plt.xlabel('Position')
plt.ylabel('Positional Encoding Value')
plt.title('Example of Sinusoidal Positional Encodings')
plt.show()
```
